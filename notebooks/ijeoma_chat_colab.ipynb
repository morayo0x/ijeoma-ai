Launches the Ijeoma AI chat interface on a Colab GPU and prints a public
`gradio.live` link. Users simply ask about West African destinations; the app
finds the relevant source passage for each question and cites it in the reply.

**Prerequisites**
1. Runtime → Change runtime type → **T4 GPU**.
2. A Colab secret named `HF_TOKEN` (key icon in the left sidebar), with notebook
   access enabled, for a Hugging Face account that has accepted the
   [Gemma licence](https://huggingface.co/google/gemma-3-1b-it).

Run all cells. The link is printed in the last cell after three to five minutes
and remains valid for up to 72 hours or until the session ends.

```python
from google.colab import userdata
for name in ("HF_TOKEN", "IjeomaHF", "HUGGINGFACE_TOKEN"):
    try:
        token = userdata.get(name)
    except Exception:
        token = None
    if token:
        os.environ["HF_TOKEN"] = token
        print("Hugging Face token found in secret", name)
        break
else:
    raise RuntimeError("No Hugging Face token in Colab Secrets. Add HF_TOKEN and enable notebook access.")
```
